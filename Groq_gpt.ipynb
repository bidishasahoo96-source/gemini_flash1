```python
import os
from groq import Groq
from google.colab import userdata

os.environ["GROQ_API_KEY1"] = userdata.get("GROQ_API_KEY1")
```

```python
client = Groq()
```

```python
completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",   # or "openai/gpt-oss-20b" (faster, smaller)
    messages=[
        {
            "role": "user",
            "content": "give me a list of animals are going to be extincted in future"

        }
    ]
)

print(completion.choices[0].message.content)
```
